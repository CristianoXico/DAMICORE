## Execução do pipeline

```python
import re
import numpy as np

# Iterate through all columns and rows to clean strings that look like arrays
for col in df.columns:
    # Apply the cleaning only to columns that are of object type (likely containing strings)
    if df[col].dtype == 'object':
        # Remove espaços, colchetes, chaves, aspas
        df[col] = df[col].astype(str).apply(
            lambda x: re.sub(r'[\s\[\]{}"]', '', x) if isinstance(x, str) and re.match(r'[\s]*[{[]', x.strip()) else x
        )
        # Remover vírgulas duplicadas e bordas
        df[col] = df[col].astype(str).apply(
            lambda x: re.sub(r',+', ',', x).strip(',') if isinstance(x, str) else x
        )
        # 🔹 Converter "nan", "NaN", "None", "<NA>" em np.nan
        df[col] = df[col].replace(["nan", "NaN", "None", "<NA>", "null"], np.nan)

print("Removed spaces, specific characters, empty strings, and excluded nan-like strings.")
display(df)
```

```python
import pandas as pd
import numpy as np
from collections import Counter

# Identificar colunas candidatas (objetos que parecem listas)
potential_array_columns = []
for col in df.columns:
    if df[col].dtype == 'object':
        if df[col].dropna().astype(str).str.contains(',').any():
            potential_array_columns.append(col)

print(f"Identified columns for modal element processing: {potential_array_columns}")

categorization_legends = {}
df_categorized = df.copy()

for col in potential_array_columns:
    print(f"\nProcessing column: {col}")

    modal_elements_list = []

    for _, row in df.iterrows():
        val = row[col]

        if isinstance(val, str):
            elements = [e.strip() for e in val.split(',') if e.strip()]
            if elements:
                counts = Counter(elements)
                modal_element = counts.most_common(1)[0][0]
                modal_elements_list.append(modal_element)
            else:
                modal_elements_list.append(np.nan)
        elif pd.notna(val):
            modal_elements_list.append(val)
        else:
            modal_elements_list.append(np.nan)

    # Criar série de elementos modais
    modal_elements_series = pd.Series(modal_elements_list, index=df.index)

    # 🔹 Padronizar valores "nan" (string) como np.nan
    modal_elements_series = modal_elements_series.replace(["nan", "NaN", "None", ""], np.nan)

    # 🔹 Frequências sem NaN
    modal_element_counts_series = modal_elements_series.dropna().value_counts()

    # Ordenar os elementos do menos frequente para o mais frequente
    sorted_unique_modal_elements = modal_element_counts_series.sort_values(ascending=True).index.tolist()

    # 🔹 Criar mapping sem NaN
    legend_mapping = {modal_value: code for code, modal_value in enumerate(sorted_unique_modal_elements)}

    # Guardar a legenda
    categorization_legends[col] = legend_mapping

    # Aplicar mapping (NaN continua como NaN no DataFrame final)
    df_categorized.loc[:, col] = modal_elements_series.map(legend_mapping)

    # --- Impressão ---
    print(f"Column '{col}':")
    if len(legend_mapping) == 0:
        print("{}")
    else:
        print("{", end="")
        for i, (key, value) in enumerate(legend_mapping.items()):
            freq = modal_element_counts_series.get(key, 0)
            print(f"'{key}': {value} ({int(freq)})", end="")
            if i < len(legend_mapping) - 1:
                print(", ", end="")
        print("}")
```

```python
#sem nan nas categorias

import json

# Arquivo de saída
output_filename = 'no_nan_categorization_' + file_name1 + '.json'

# Estrutura de dicionário para exportar em JSON
export_legends = {}

for col, legend in categorization_legends.items():
    # Frequências do modo já processado
    modal_elements_series = df_categorized[col].map({v: k for k, v in legend.items()})
    modal_element_counts_series = modal_elements_series.value_counts(dropna=False)

    col_dict = {}
    for key, code in sorted(legend.items(), key=lambda item: item[1]):
        # Exclude 'nan' category from the JSON output
        if not pd.isna(key):
            freq = modal_element_counts_series.get(key, 0)
            col_dict[str(key)] = {"code": code, "frequency": int(freq)}

    export_legends[col] = col_dict

# Salvar em JSON (indentado para legibilidade)
with open(output_filename, 'w', encoding='utf-8') as f:
    json.dump(export_legends, f, ensure_ascii=False, indent=2)

print(f"Categorization legends with frequencies saved to '{output_filename}' (excluding 'nan' category)")
```

## Get user input for objectives


### Subtask:
Prompt the user to select the columns to be used as objectives and whether to minimize or maximize each.


**Reasoning**:
Display the available columns to the user and prompt for objective column selection and optimization direction, including input validation.



```python
available_columns = df_categorized.columns.tolist()
print("Available columns for objectives:")
for i, col in enumerate(available_columns):
    print(f"{i+1}. {col}")

selected_objectives = {}
while True:
    objective_input = input("Enter the names of the columns to use as objectives, separated by commas: ")
    objective_names = [name.strip() for name in objective_input.split(',')]

    valid_columns = True
    for name in objective_names:
        if name not in available_columns:
            print(f"Invalid column name: {name}. Please select from the available columns.")
            valid_columns = False
            break

    if valid_columns:
        for name in objective_names:
            while True:
                direction_input = input(f"Optimize '{name}' for minimization or maximization? (enter 'min' or 'max'): ").strip().lower()
                if direction_input in ['min', 'max']:
                    selected_objectives[name] = direction_input
                    break
                else:
                    print("Invalid optimization direction. Please enter 'min' or 'max'.")
        break

print("\nSelected objectives and optimization directions:")
print(selected_objectives)
```

```python
from pymoo.core.problem import Problem
from pymoo.algorithms.moo.nsga2 import NSGA2
from pymoo.optimize import minimize
from pymoo.util.nds.non_dominated_sorting import NonDominatedSorting
import numpy as np
import pandas as pd
from typing import List


# Define a custom Problem class for NSGA-II
class MyProblem(Problem):

    def __init__(self, objective_data, selected_objectives_keys):
        # objective_data is a NumPy array of the objective values
        # selected_objectives_keys is a list of column names to use as objectives

        # Number of variables (features) is the number of columns in the objective data
        n_var = objective_data.shape[1]

        # Number of objectives is the number of selected columns
        n_obj = objective_data.shape[1]

        # Define lower and upper bounds for variables (using the min/max values from the objective data)
        # Add a small epsilon to the upper bound to avoid issues with floating point comparisons during optimization
        epsilon = 1e-6
        # Use nanmin and nanmax to handle potential NaNs in the initial data before imputation
        xl = np.nanmin(objective_data, axis=0)
        xu = np.nanmax(objective_data, axis=0) + epsilon


        # NSGA-II minimizes by default. We will handle maximization by negating objectives in _evaluate.
        super().__init__(n_var=n_var, n_obj=n_obj, xl=xl, xu=xu, elementwise_evaluation=True)

        self.objective_data = objective_data # Store the objective data
        self.selected_objectives_keys = selected_objectives_keys # Store the selected objective keys

    def _evaluate(self, x, out, *args, **kwargs):
        # x represents a single solution (a row from objective_data_np in this setup)
        # The objective values are the values in the row itself

        f = np.copy(x) # Start with the objective values

        # Adjust objective values based on optimization direction
        # Retrieve the optimization direction for each objective using the stored keys
        objective_directions = [selected_objectives[key] for key in self.selected_objectives_keys]

        for i, direction in enumerate(objective_directions):
            if direction == 'max':
                f[i] *= -1 # Negate for maximization (NSGA-II minimizes by default)

        out["F"] = f


# --- Prepare the objective data using selected_objectives ---
# Select the columns that were chosen as objectives
objective_columns = list(selected_objectives.keys())
objective_data_df = df_categorized[objective_columns].copy() # Use .copy() to avoid SettingWithCopyWarning

# Replace empty strings with NaN and convert to numeric
objective_data_df.replace('', np.nan, inplace=True)
for col in objective_data_df.columns:
    objective_data_df[col] = pd.to_numeric(objective_data_df[col], errors='coerce')

# --- Check for and exclude columns with all NaN values ---
cols_to_exclude = objective_data_df.columns[objective_data_df.isnull().all()].tolist()
if cols_to_exclude:
    print(f"Excluding objective columns with all NaN values: {cols_to_exclude}")
    objective_columns_for_nsga2 = [col for col in objective_columns if col not in cols_to_exclude]
    objective_data_df_for_nsga2 = objective_data_df.drop(columns=cols_to_exclude)
else:
    objective_columns_for_nsga2 = objective_columns
    objective_data_df_for_nsga2 = objective_data_df.copy() # Ensure a copy is used

# --- Handle missing values (NaNs) by imputation instead of dropping rows ---
# Impute missing values with the mean of the column
if not objective_data_df_for_nsga2.empty:
    objective_data_df_for_nsga2.fillna(objective_data_df_for_nsga2.mean(), inplace=True)
# --- End of imputation ---

# Now convert the cleaned and imputed data to a NumPy array
objective_data_np = objective_data_df_for_nsga2.values.astype(float)


# --- Run NSGA-II to get ranks ---
if objective_data_np.shape[0] > 0 and objective_data_np.shape[1] > 0: # Ensure there are samples and objectives left after preparation
    # Instantiate the custom problem
    problem = MyProblem(objective_data_np, objective_columns_for_nsga2) # Pass the list of keys used for NSGA-II

    # Configure the NSGA-II algorithm
    # Pop size should ideally be a multiple of 4 for NSGA-II
    pop_size = 100
    if objective_data_np.shape[0] < pop_size:
        pop_size = objective_data_np.shape[0] # Adjust pop_size if data is smaller

    algorithm = NSGA2(pop_size=pop_size)

    # Run the algorithm
    # Number of generations can be adjusted
    n_gen = 100
    if objective_data_np.shape[0] < 100: # Adjust generations if data is very small
         n_gen = 50


    print("\nRunning NSGA-II algorithm to calculate ranks...")
    # We run the algorithm on the entire dataset to get ranks for all samples
    results = minimize(problem,
                       algorithm,
                       ('n_gen', n_gen),
                       seed=1,
                       verbose=True)

    print("\nNSGA-II algorithm finished.")
    print("Results object:", results)

    # --- Get NSGA-II ranks for all solutions ---
    # The results.F contains the objective values of the solutions in the last generation's population
    # We need to perform non-dominated sorting on these objective values to get the ranks
    # However, pymoo's minimize function already provides the final population's objective values in results.F
    # and the corresponding solutions (decision variables) in results.X.
    # The non_dominated_sorting function from pymoo.util.nds can be used to get the fronts.

    # Apply NonDominatedSorting to the original (imputed) objective data
    # First, prepare the objective data for sorting (negate for maximization)
    objective_data_for_sorting = objective_data_np.copy()
    # Adjust objective values for sorting based on optimization direction for the columns used in NSGA-II
    objective_directions_for_nsga2 = [selected_objectives[key] for key in objective_columns_for_nsga2]
    for i, direction in enumerate(objective_directions_for_nsga2):
        if direction == 'max':
            objective_data_for_sorting[:, i] *= -1 # Negate for maximization

    # Perform non-dominated sorting on the entire dataset
    fronts_all_data = NonDominatedSorting().do(objective_data_for_sorting)

    # Create an array to store ranks for all data points
    nsga_ii_ranks_all_data = np.zeros(objective_data_for_sorting.shape[0], dtype=int)

    # Assign ranks based on the fronts, starting from 1
    for rank, front in enumerate(fronts_all_data):
        nsga_ii_ranks_all_data[front] = rank + 1 # Start rank from 1


    # Add the NSGA-II rank as a new column to the df_categorized DataFrame
    # Ensure the ranks are assigned to the correct original rows
    # Since we performed sorting on objective_data_for_sorting which is derived from
    # objective_data_np (which has the same index as objective_data_df and df_categorized),
    # the ranks correspond directly to the rows in df_categorized.
    df_categorized['nsga_ii_rank'] = nsga_ii_ranks_all_data

    # Create the final DataFrame with spatial_geo_code, selected objectives, and NSGA-II rank
    # Include all originally selected objective columns in the output_df, even if excluded from NSGA-II
    output_df = df_categorized[['spatial_geo_code'] + objective_columns + ['nsga_ii_rank']].copy()

    print("\nDataFrame with spatial_geo_code, selected objectives, and NSGA-II rank:")
    display(output_df.head())

    print("\nShape of the output DataFrame:", output_df.shape)

    # Define the output filename for the ranked data CSV
    output_ranked_filename = 'no_nan_ranked_data_nsgaii_' + file_name1 + '.csv'

    # Save the output DataFrame to a CSV file
    if not output_df.empty:
        output_df.to_csv(output_ranked_filename, index=False)
        print(f"\nRanked data saved to '{output_ranked_filename}'")
    else:
        print("\nNo ranked data to save.")


else:
    print("\nNo data points or objectives available for NSGA-II ranking after preparation.")
    # Create an empty DataFrame with all originally selected objective columns and the rank column
    output_df = pd.DataFrame(columns=['spatial_geo_code'] + objective_columns + ['nsga_ii_rank'])
    print("\nEmpty DataFrame created as no data or objectives were available for ranking.")

# Note: The output_df variable now holds the spatial_geo_code, selected objectives, and NSGA-II rank for each sample.
```
