```python
import geopandas as gpd

# Assuming all shapefile components are in the same directory as 'BR_Regioes_2024.shp'
file_path = 'br/RG2017_regioesgeograficas2017.shp'
mapa = gpd.read_file(file_path)

# Display the first few rows of the GeoDataFrame
display(mapa.head())
```

```python
# Filter the GeoDataFrame to select municipalities in the state of São Paulo (SP)
mapa_sp = mapa[mapa['UF'] == '35']

# Display the first few rows of the filtered GeoDataFrame
display(mapa_sp)
```

```python
mapa.columns
```

```python
mapa_sp.plot(color='white', edgecolor='black')
```

```python
# Simplify the geometries in the GeoDataFrame
# You can adjust the 'tolerance' parameter to control the level of simplification.
# A larger tolerance means more simplification.
mapa_simplificado = mapa.simplify(tolerance=0.03)

# Display the simplified GeoDataFrame (optional)
# display(mapa_simplificado.head())

# Plot the simplified GeoDataFrame to visualize the result
mapa_simplificado.plot(color='white', edgecolor='black')
```

```python
import folium

# Get the bounds of the simplified GeoDataFrame to center the map
bounds = mapa_simplificado.total_bounds
center_lat = (bounds[1] + bounds[3]) / 2
center_lon = (bounds[0] + bounds[2]) / 2

# Create a Folium map centered on the data
m = folium.Map(location=[center_lat, center_lon], zoom_start=4)

# Add the simplified GeoDataFrame to the map
folium.GeoJson(mapa_simplificado).add_to(m)

# Display the map
m
```

```python
display(mapa_simplificado.head())
```

```python
# Ensure the indices are aligned
mapa['CD_MUN'] = mapa['CD_MUN']

# Add the 'CD_MUN' column from the original mapa GeoDataFrame to mapa_simplificado
# We can do this by aligning the indices, assuming the order is the same
mapa_simplificado = mapa_simplificado.to_frame().join(mapa['CD_MUN'])

# Display the first few rows of the modified mapa_simplificado GeoDataFrame
display(mapa_simplificado.head())
```

```python
# Save the GeoDataFrame as a GeoJSON file
mapa_simplificado.to_file(file_name + ".geojson", driver='GeoJSON')

print("GeoDataFrame saved as mapa.geojson")
```
