# Task
Create a pandas DataFrame from the JSON data at "https://servicodados.ibge.gov.br/api/v1/localidades/municipios?orderBy=nome", where each row represents a municipality and includes all nested fields as columns. Export the DataFrame to a CSV file.

Or use the Query Builder for get your map: "https://servicodados.ibge.gov.br/api/docs/malhas?versao=4#api-bq"

## Fetch and load data


```python
import requests

url = "https://servicodados.ibge.gov.br/api/v1/localidades/municipios?orderBy=nome"
response = requests.get(url)
response.raise_for_status()
data = response.json()
```

## Normalize json data


```python
import pandas as pd

normalized_data = pd.json_normalize(data)
```

```python
import pandas as pd

normalized_data = pd.json_normalize(data)
```

## Export to csv


```python
normalized_data.to_csv('municipios.csv', index=False)
```
